```python
# 🔷 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# 🔷 2. Load Dataset
df = pd.read_csv("C:/Users/saidh/Downloads/FACT_LAB_TEST_RESULT.csv")

# 🔷 3. Clean Column Names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# 🔷 4. Count Normal vs Abnormal
counts = df['normal_flag'].value_counts()

print("Lab Test Distribution:\n", counts)

# 🔷 5. Bar Chart with Colors
colors = ['green' if label == 'Normal' else 'red' for label in counts.index]

plt.figure()
counts.plot(kind='bar', color=colors)
plt.title("Lab Test Results Distribution")
plt.xlabel("Test Result")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()

# 🔷 6. Pie Chart with Colors
plt.figure()
counts.plot(kind='pie', autopct='%1.1f%%', colors=['green','red'])
plt.title("Normal vs Abnormal Lab Results")
plt.ylabel("")
plt.show()
```

```python
# 🔷 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# 🔷 2. Load Dataset
df = pd.read_csv("C:/Users/saidh/Downloads/FACT_INSURANCE_CLAIM.csv")

# 🔷 3. Clean Column Names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# 🔷 4. Separate Approved & Rejected Claims
approved = df[df['claim_status'] == 'Approved']['claim_amount']
rejected = df[df['claim_status'] == 'Rejected']['claim_amount']

# 🔷 5. Histogram (Combined View)
plt.figure()

plt.hist(approved, bins=10, alpha=0.7, label='Approved')
plt.hist(rejected, bins=10, alpha=0.7, label='Rejected')

plt.title("Insurance Claim Amount Distribution by Status")
plt.xlabel("Claim Amount")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

```python
# 🔷 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# 🔷 2. Load Dataset
df = pd.read_csv("C:/Users/saidh/Downloads/FACT_PHARMACY_DISPENSE.csv")

# 🔷 3. Clean Column Names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# 🔷 4. Total Quantity per Drug
drug_usage = df.groupby('drug_id')['quantity'].sum().sort_values(ascending=False)

print(drug_usage.head())

# 🔷 5. Bar Chart (Top 10 Drugs)
plt.figure()
drug_usage.head(10).plot(kind='bar')

plt.title("Top 10 Most Dispensed Drugs")
plt.xlabel("Drug ID")
plt.ylabel("Total Quantity")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("C:/Users/saidh/Downloads/FACT_TREATMENT_EVENT.csv")

# Create box plot
plt.figure(figsize=(8,5))

plt.boxplot(df['treatment_cost'],
            patch_artist=True,   # enables color
            boxprops=dict(facecolor='lightblue'),
            medianprops=dict(color='red'),
            whiskerprops=dict(color='black'),
            capprops=dict(color='black'))

plt.title("Distribution of Treatment Costs")
plt.ylabel("Treatment Cost")

plt.show()
```
